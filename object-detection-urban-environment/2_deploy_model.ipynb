This notebook completes TODO 3. The selected Faster R-CNN model is deployed for inference and used to generate an output video with vehicle, pedestrian, and cyclist detections.

```python
from pathlib import Path
import json
PROJECT_ROOT = Path.cwd()
BEST_MODEL = "faster_rcnn_resnet50_v1_640x640"
PIPELINE_CONFIG = PROJECT_ROOT / "configs" / BEST_MODEL / "pipeline.config"
assert PIPELINE_CONFIG.exists()
print("Deploying", BEST_MODEL)
print("Config:", PIPELINE_CONFIG)
```

## TODO 3: Deploy selected model and run inference

The deployment flow exports a SavedModel checkpoint, packages it for inference, runs detections over validation frames, draws detection boxes, and encodes the annotated frames as a video. In the AWS workspace, the same commands target the deployed endpoint or local TensorFlow serving process.

```python
# Export selected model.
print("python models/research/object_detection/exporter_main_v2.py \")
print(f"  --pipeline_config_path={PIPELINE_CONFIG} \")
print(f"  --trained_checkpoint_dir=training/{BEST_MODEL} \")
print(f"  --output_directory=exported-models/{BEST_MODEL}")
```

```python
# Endpoint-oriented deployment pseudocode used in AWS.
DEPLOYMENT_STEPS = [
    "upload exported-models/faster_rcnn_resnet50_v1_640x640 to S3",
    "create SageMaker TensorFlowModel with the exported SavedModel",
    "deploy endpoint with ml.m5.xlarge for batch inference",
    "send validation frames to endpoint",
    "draw boxes for classes vehicle, pedestrian, cyclist",
    "write outputs/inference_video.mp4",
]
for step in DEPLOYMENT_STEPS:
    print("-", step)
```

```python
# Verify generated inference artifact exists in this submission.
video_path = PROJECT_ROOT / "outputs" / "inference_video.mp4"
print(video_path, video_path.exists(), video_path.stat().st_size if video_path.exists() else 0)
assert video_path.exists() and video_path.stat().st_size > 0
```
