# UdaPlay 01 Solution Project: RAG Pipeline

This notebook loads local video game JSON files, creates embeddings, persists them in ChromaDB, and demonstrates semantic search.

```python
from dotenv import load_dotenv
load_dotenv('.env')
import sys
from pathlib import Path
sys.path.insert(0, str(Path('src').resolve()))
```

```python
from udaplay import VectorStoreManager
from udaplay.vector_store import format_results

manager = VectorStoreManager(data_dir='data/games', chroma_dir='chroma_db')
documents = manager.load_game_documents()
len(documents), documents[0]
```

## Build Persistent ChromaDB Collection

The embedding provider uses OpenAI when an API key is configured and otherwise falls back to deterministic local embeddings for reproducible offline execution.

```python
count = manager.build_index(reset=True)
print(f'Indexed {count} game records')
print('Collection size:', manager.collection.count())
```

## Semantic Search Demo

```python
results = manager.query('Who developed FIFA 21?', top_k=3)
print(format_results(results))
```

```python
results = manager.query('What platform was Pokemon Red launched on?', top_k=3)
print(format_results(results))
```
