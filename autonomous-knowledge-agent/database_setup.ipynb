# UDA-Hub Database Setup

This notebook initializes the SQLite database used by the Autonomous Knowledge Agent. It creates the required `Account`, `User`, `Ticket`, `TicketMetadata`, `TicketMessage`, and `Knowledge` tables, plus persistent memory and tool logging tables used by the workflow.

```python
from database import DB_PATH, initialize_database, get_connection

db_path = initialize_database(DB_PATH, reset=True)
db_path
```

```python
tables = ["Account", "User", "Ticket", "TicketMetadata", "TicketMessage", "Knowledge", "CustomerMemory", "ToolLog"]
with get_connection(db_path) as conn:
    for table in tables:
        count = conn.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0]
        print(f"{table}: {count}")
```

The `Knowledge` table is seeded with 15 articles, satisfying the requirement for at least 14 knowledge articles.
