# 01 Database Setup

This notebook initializes the EcoHome SQLite database and populates it with reproducible sample data for energy usage and solar generation. The project rubric requires the database to contain `EnergyUsage` and `SolarGeneration` tables before the agent is evaluated.

```python
from models.energy import initialize_database, populate_sample_data, query_rows
initialize_database()
populate_sample_data()
print('EnergyUsage rows:', query_rows('SELECT COUNT(*) AS count FROM EnergyUsage')[0]['count'])
print('SolarGeneration rows:', query_rows('SELECT COUNT(*) AS count FROM SolarGeneration')[0]['count'])
```

## Energy Usage Summary

This table shows historical device-level usage and cost totals used by the recommendation agent.

```python
query_rows('SELECT device_type, ROUND(SUM(consumption_kwh), 2) AS kwh, ROUND(SUM(cost_usd), 2) AS cost FROM EnergyUsage GROUP BY device_type ORDER BY cost DESC')
```

## Solar Generation Summary

This aggregation verifies that weather-conditioned solar generation data exists for analysis.

```python
query_rows('SELECT weather_condition, ROUND(AVG(generation_kwh), 2) AS avg_generation FROM SolarGeneration GROUP BY weather_condition')
```
